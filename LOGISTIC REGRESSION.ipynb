 **1. What is Simple Linear Regression (SLR)? Explain its purpose.**

 --> Simple Linear Regression (SLR) shows the relationship between one independent variable (X) and one dependent variable (Y)

It uses a straight line equation to model the relationship

It helps understand, explain, and predict values of Y based on X



**2. What are the key assumptions of Simple Linear Regression?**

--> Key assumptions of Simple Linear Regression:

* Linearity – The relationship between X and Y is linear

* Independence – Observations are independent of each other

* Homoscedasticity – Constant variance of errors across all values of X

* Normality – Errors are normally distributed

* No multicollinearity – Only one independent variable is used (automatically satisfied in SLR)

**3. Write the mathematical equation for a simple linear regression model and explain each term.**

--> Simple Linear Regression Model

𝑌 = 𝛽0 + 𝛽1𝑋 + 𝜀

𝑌 – Dependent variable (response/output)

𝑋 – Independent variable (predictor/input)

𝛽0 – Intercept (value of Y when X = 0)

𝛽1 – Regression coefficient or slope (change in Y for one-unit change in X)

𝜀 – Error term (accounts for randomness and unexplained variation)

**5. What is the method of least squares in linear regression?**

--> **Method of Least Squares:**

* A technique to find the best-fitting line in linear regression.

* Minimizes the sum of squared differences between actual values and predicted values.

* Ensures the regression line fits the data as closely as possible.

**6. What is Logistic Regression? How does it differ from Linear Regression?**

--> **Logistic Regression**

Logistic regression is used to predict a categorical outcome (usually binary, like Yes/No or 0/1) based on one or more independent variables. It estimates the probability of an event using the logistic (sigmoid) function.

**Difference from Linear Regression**

Linear regression predicts a continuous value, while logistic regression predicts a probability or class. Linear regression uses a straight-line equation, whereas logistic regression uses a sigmoid curve to ensure the output is between 0 and 1.

**7. Name and briefly describe three common evaluation metrics for regression models.**

--> **Three Common Evaluation Metrics for Regression Models**

**1. Mean Absolute Error (MAE)**

* Measures the average absolute difference between actual and predicted values.
* Lower MAE indicates better predictions.

**2. Mean Squared Error (MSE)**

* Measures the average squared difference between actual and predicted values.
* Penalizes larger errors more than MAE.

**3. R-squared (𝑅2)**
* Represents the proportion of variance in the dependent variable explained by the model.
* Values range from 0 to 1; closer to 1 means better fit.

**8. What is the purpose of the R-squared metric in regression analysis?**

--> **Purpose of R-squared in Regression**

* **Measures goodness of fit:** R-squared shows how well the regression model explains the variation in the dependent variable.

* **Value range:** Between 0 and 1 (0 = model explains nothing, 1 = model explains all variation).

* **Interpretation:** Higher R-squared means the model fits the data better and predicts more accurately.

**In short:** It tells us how much of the variation in Y is explained by X.

```python
# 9. Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.

from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([50, 55, 65, 70, 75])

model = LinearRegression()
model.fit(X, Y)

print("Slope (β1):", model.coef_[0])
print("Intercept (β0):", model.intercept_)
```

**10. How do you interpret the coefficients in a simple linear regression model?**

-->

**1. Intercept (𝛽0):**

* The value of the dependent variable (Y) when the independent variable (X) is zero.

* It represents the starting point of the regression line.

**Slope (𝛽1):**

* The amount by which Y changes for a one-unit change in X.

* It shows the strength and direction of the relationship between X and Y.

**Summary:**

* Intercept = baseline value of Y

* Slope = effect of X on Y
